**4- Dates and Times**

Python has a datetime module built into the standard library, which offers tools for working with dates.

In real datasets, you'll find:

    Different date formats

    Missing time zones

    Weird offsets

    Inconsistent timestamps

    Business logic like "last 30 days" or "next business week"

... and it's up to you to figure out how to handle it.

**Parsing Timestamps**

You'll often receive a timestamp as a string:

    "2026-05-28 14:30:00"
we convert from a **string** to a **datetime object** using datetime.strptime()

The timestamp here is an actual datetime object, not just a string. You can do math with it, compare it to other dates, etc.

That format string, "%Y-%m-%d %H:%M:%S", tells Python how to parse the text as a date and time:

    %Y = 4-digit year

    %m = month

    %d = day

    %H = hour in 24-hour time

    %M = minute

    %S = second

```python
from datetime import datetime
```

```python
timestamp_str = "2026-05-28 14:30:00"
timestamp_str
```

```python
timestamp_obj = datetime.strptime(timestamp_str, "%Y-%m-%d %H:%M:%S")
timestamp_obj
```

```python
print(timestamp_obj)
```

```python
#ex2
iso_date = datetime.strptime("2024-01-15", "%Y-%m-%d")
iso_date
```

```python
#ex3
us_date = datetime.strptime("01/15/2024", "%m/%d/%Y")
us_date
```

```python
eu_date = datetime.strptime("15/01/2024", "%d/%m/%Y")
eu_date
```

**5- Datetime Math**

Once you've parsed a date or timestamp, you usually want to do something with it
That's where **timedelta** comes in

**Subtract** one datetime from another and you get a timedelta

```python
order_placed = datetime(2026, 5, 25, 14, 30, 0)
delivered_at = datetime(2026, 5, 29, 14, 30, 0)
difference = delivered_at - order_placed
difference
```

**Adding or subtracting timedelta from time**

You can also add or subtract a **timedelta** from a **datetime** to get a new datetime:

```python
from datetime import timedelta
```

```python
order_placed = datetime(2026, 5, 25, 14, 30, 0)
expected_by_later = order_placed + timedelta(hours=6)
expected_by_later
```

```python
order_placed = datetime(2026, 5, 25, 14, 30, 0)
expected_by_early = order_placed - timedelta(days=6)
expected_by_early
```

**6-1- Basic Comparisons**

You can compare two datetime values with operators like <, >, <=, and >=:

```python
returned_at = datetime(2026, 5, 15)
due_date = datetime(2026, 6, 1)

if returned_at < due_date:
    print("Book returned on time")
```

**7- List Comprehensions**

In data work, you're constantly building new lists. Say you have a list of songs:

    tracks = [

        {"track_id": "T-100", "plays": 2},

        {"track_id": "T-101", "plays": 5},

        {"track_id": "T-102", "plays": 4},
    ]

```python
tracks = [
    {"track_id": "T-100", "plays": 2},
    {"track_id": "T-101", "plays": 5},
    {"track_id": "T-102", "plays": 4},
]
tracks
```

```python
# And you want a new list with just the number of "plays" for each track.
# You can do that with a normal for loop:
play_counuts = []
for track in tracks:
    play_counuts.append(track["plays"])
print(play_counuts)
```

a shorter way to do it in python
**The syntax is**:

[expression for item in iterable]

```python
play_counts = [track["plays"] for track in tracks]
play_counts
```

And if you want to filter as well:

[expression for item in iterable if condition]

```python
play_counts = [track["plays"] for track in tracks if track["plays"] > 2]
play_counts
```

Dictionary Comprehensions

If list comprehensions build lists, dictionary comprehensions build dictionaries.

meaning to go from

    {"game_id": "g-01", "title": "The Legend of Zelda", "status": "released"}

to

    'g-01': 'The Legend of Zelda'
the syntax is

{key_expression: value_expression for item in iterable}

we have

    games = [
      {"game_id": "g-01", "title": "The Legend of Zelda", "status": "released"},
      {"game_id": "g-02", "title": "Half Life 3", "status": "delayed"},
      {"game_id": "g-03", "title": "Hades II", "status": "released"},
    ]

and want a dictionary where you can lookup game_id --> title. You can use a dictionary comprehension:

```python
games = [
  {"game_id": "g-01", "title": "The Legend of Zelda", "status": "released"},
  {"game_id": "g-02", "title": "Half Life 3", "status": "delayed"},
  {"game_id": "g-03", "title": "Hades II", "status": "released"},
]
games
```

```python
game_titles = {game["game_id"]: game["title"] for game in games}
game_titles
```

To filter records while creating the dictionary, you can optionally add an if condition

the syntax is:

    {key_expression: value_expression for item in iterable if condition}

```python
relsesed_titles = {game["game_id"]: game["title"] for game in games if game["status"] == "released"}
relsesed_titles
```
